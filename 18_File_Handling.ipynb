# File Handling

* File Handling means creating, reading, updating, deleting (CRUD) operations that we can perform in files.
* We have to use open() function to open a file in python.

```python
"""
Syntax:-
    file = open("file path", "mode")
    print(file.read())            # Read entire file
    print(file.readline())        # Read one line
    print(file.readlines())       # Read all lines into a list

    file.close()
"""
```

> There are multiple modes to open the file.
1. 'r' : Read(default) - file must exist.
2. 'w' : Write - creates file or overwrites.
3. 'a' : Append - adds to end of file.
4. 'x' : Create - created a new file, fails if it exists.

```python
r = open('spiderman.txt','w')
```

```python
r.write("Hello, I am Zaid and we are learning Python")
r.close()
```

```python
r=open('spiderman.txt','a')
r.write("Now i am using appending mode")
r.close()
```
