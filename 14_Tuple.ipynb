# Tuple

- A tuple is an ordered, immutable collection of items. It can store elements of different data types, and duplicate values are allowed.

```python
""" 
Syntax:-
    tuple_name = (item1, item2, item3)
    
"""
```

### Features of a Tuple
- Ordered – Items maintain their insertion order.
- Immutable – Items cannot be changed after creation.
- Allows duplicates – Duplicate values are allowed.
- Can store different data types – Integers, strings, floats, lists, etc.

```python
num = (1,4,2,6)
print(type(num))
```

#### Traversing in Tuple  
Same as in List

#### Methods of Tuple

```python
num=(12,4,3,2.5,'Hello',3)

index = num.index(3)    # Finds the index of first occurence of 3
count_n = num.count(3)     # Counts total occurences of 3
```

```python
print(index)
print(count_n)
```

### Tuple Unpacking

```python
a,b,c,d=(1,2,3,4)
print(a)
print(b)
print(c)
print(d)
```

>Note  
- If you want to give only one element in tuple, you have to provide a comma after it otherwise it will be consider as int.

```python
## Example-

# Without Comma
a=(1)
print(type(a))

# With Comma
b=(1,)
print(type(b))
```

### Advantages of Tuple  
- Faster than lists for storing fixed data.
- Prevents accidental modification of data.
- Can be used as dictionary keys (if all elements are immutable).
- Uses less memory than lists.
