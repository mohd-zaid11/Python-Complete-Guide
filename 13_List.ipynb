# List

- A list is an ordered, mutable collection of items. It can store elements of different data types, and duplicate values are allowed.

```python
"""
Syntax for creating list:-
    list_name = [item1, item2, item3]
"""
```

### Features of a List
- Ordered – Items maintain their insertion order.
- Mutable – Items can be added, removed, or modified.
- Allows duplicates – Duplicate values are allowed.
- Heterogenous – Can store multiple data types like Integers, strings, floats, lists, etc.

```python
lst = [1, 2 , 6.2 , 2 , 'Zaid' , 3-4j, True]
print(type(lst))
print(lst)
```

-> Slicing in List works same as in String

```python
print(lst[2])
print(lst[2:5])
print(lst[0:8:2])
```

#### Traversing in List

```python
## Using Index

lst = [1, 2 , 6.2 , 2 , 'Zaid' , 3-4j, True]
for i in range(len(lst)):
    print(lst[i])
```

```python
## Direclty on values

lst = [1, 2 , 6.2 , 2 , 'Zaid' , 3-4j, True]
for i in lst:
    print(i)
```

### Some List Methods

```python
num = [1,5,2,9,5,6]

num.append(10)  #Adds 10 to the end
num.insert(2,12)  #Adds 12 at index 2
num.extend([15,20,25])  #Adds multiple elements at the end
num.remove(5)   #Removes the first occurence of 5
popped_item = num.pop(3)    #Renoves and stores the element as index 3
index = num.index(2)    #Finds the index of 2
count_5 = num.count(5)    #Counts occurence of 5
num.sort()     #Sorts the list in ascending order
num.reverse()    #Reverse the list order
new_num = num.copy()    #Creates a copy of the list
num.clear()     #Removes all the elements from the lsit
```

### Practice Questions on List

```python
## Print positive and negative elements of an List
num=[11,2,8,-4,1,-8,0,-5,7]

for i in num:
    if i>0:
        print(f"{i} is positive")
    elif i<0:
        print(f"{i} is negative")
    else:
        print(i)
```

```python
## Mean of List elements

num=[3,2,9,4,5,7,7]
sum=0
for i in num:
    sum+=i
mean=sum/len(num)
print(f"Mean of list is {mean}")
```

```python
## Find greatest element an print its index too

num=[3,2,9,12,4,16,5,10,7,7,18]

largest=num[0]
for i in range(len(num)):
    if num[i]>largest:
        largest=num[i]
        index=i

print(f"Largest element is {largest} at index {index}")
```

```python
## Find the second greatest number

num=[3,2,9,12,4,16,5,10,7,7,18,17]

largest = num[0]
sec_largest=num[0]
for i in num:
    if i>largest:
        sec_largest=largest
        largest=i
    elif i>sec_largest:
        sec_largest=i

print("Second largest is", sec_largest)
```
