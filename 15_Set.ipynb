# Set

- A set is an unordered, mutable collection of unique items. It automatically removes duplicate values.

```python
""" 
Syntax:-
    set_name = {item1, item2, item3}
    
"""
```

### Features of a Set
- Unordered – Items do not have a fixed order.
- Mutable – Items can be added or removed.
- Unique – Duplicate values are automatically removed.
- Can store different data types – Integers, strings, tuples, etc.  

Set uses hash value to store values.

```python
num = {1,4,2,6,5,9}
print(type(num))
```

### Traversing in Set  
Set can not be traversed using the index values cause it is unordered and has no index.  


We can use loop for traversing but it will give random value, as the values in set are unordered.

```python
a = {1,7,2,3,4,5,6}
b={'a','b','c','d'}
for i in a:
    print(i)
for i in b:
    print(i)
```

### Set Methods


```python
s = {1,2,3}

s.add(4)                    #Adds an element to the set
s.remove(2)                 #Removes 2 - Raises an error if not found
s.discard(2)                #Removes 2 - No error if not found
popped_element = s.pop()    #Removes a random element   
s.clear()                   #Removes all elements
```

> So there are some other operations of sets that can be performed between two sets. And we also have shortcuts fro them

```python
A = {1,2,3}
B = {3,4,5}

union_set = A.union(B)                      #{1,2,3,4,5}
#Shortcut:-
union_set = A | B

intersection_set = A.intersection(B)        #{3}
#Shortcut:-
intersection_set = A & B

difference_set = A.difference(B)            #{1,2}
#Shortcut:-
difference_set = A - B

symmetric_diff = A.symmetric_difference(B)  #{1,2,4,5}
#Shortcut:-
symmetric_diff = A ^ B
```
