# Loops

- Loops in Python allows us to execute a block of code multiple times without rewriting it.  


#### Types of Loops  
1. For Loop  
2. While Loop

If we know exactly number of iteration we will use For Loop otherwise, While Loop.  
For loop works on basis of number/range, While loop works on condition basis. 

## For Loop

- range(start,stop,steps)  
- range() function is used to provide a range to For loop.

```python
""" 
For Loop Syntax-  

for i in range():  
    code block

"""
```

```python
for i in range(1,10):
    print(i)
```

```python
for i in range(0,11,2):
    print(i)
```

```python
for i in range(6):
    print(i)
```

```python
for i in range(6,1,-1):
    print(i)
```

```python
for i in range(-3,-9,-1):
    print(i)
```

We can use use loop fro strings in 2 ways-  
1. using index value
2. iterating directly over the string

```python
a="Zaid"

for i in range(0,len(a)):
    print(a[i])
```

```python
a="Zaid"
for i in a:
    print(i)
```

## Break-Continue-Else

### Break  
-> The break statement is used to immediately terminate a loop when a specified condition is met.  

### Continue  
-> The continue statement skips the current iteration of a loop and moves to the next iteration.  

### Else  
-> The else block of a loop executes only if the loop completes normally without encountering a break statement.

```python
## break
for i in range(6):
    print(i)
    if i==3:
        break
    
```

```python
for i in range(6):
    if i==3:
        break
    else:
        print(i)
```

```python
## continue
for i in range(6):
    if i==3:
        continue
    print(i)
    
```

```python
## else
for i in range(5):
    print(i)
else:
    print("Loop Completed")
```

### For Loop Practice Questions

```python
## Accept an integer and print your name n times

n=int(input("Enter integer: "))

for i in range(n):
    print("Zaid")
```

```python
## Print natural number upto n

n=int(input("Enter integer: "))
for i in range(1,n+1):
    print(i)
```

```python
## Reverse for loop - Print n to 1 

n=int(input("Enter integer: "))
for i in range(n,0,-1):
    print(i)
```

```python
## Sum upto n numbers

n=int(input("Enter integer: "))
sum=0
for i in range(1,n+1):
    sum+=i
print(sum)
```

```python
## Factorial of a number

n=int(input("Enter integer: "))
fact=1
for i in range(1,n+1):
    fact*=i
print(fact)
```

```python
## Print the sum of all even & odd numbers in a range separately

n=int(input("Enter integer: "))
even_sum=0
odd_sum=0
for i in range(1,n+1):
    if i%2==0:
        even_sum+=i
    else:
        odd_sum+=i

print(f"Sum of even numbers: {even_sum}")
print(f"Sum of odd numbers: {odd_sum}")
```

```python
## Print all factors of a number

n=int(input("Enter integer: "))
for i in range(1,n+1):
    if n%i==0:
        print(i)
```

## While Loop

- This loop repeats a block of code as long as the condition is True.

```python
""" 
While Loop Syntax-

while condition:
    code to execute
    iteration
    
"""
```

```python
a=1

while a<=10:
    print(a)
    a+=1
```

### While Loop Practice Questions

```python
## Separate each digit of a number and print it on the new line.

n=int(input("Enter a number: "))

while n>0:
    print(n%10)
    n=n//10
```

```python
## Accept a number and print its reverse

n=int(input("Enter a number: "))
rem=0
rev=0
while n>0:
    rem = n%10
    rev = rev*10 + rem
    n=n//10 

print(rev)
```

```python
## Accept a number and check if it is palindrome or not

n=int(input("Enter a number: "))
m=n
rem=0
rev=0
while m>0:
    rem = m%10
    rev = rev*10 + rem
    m=m//10 

if n == rev:
    print("It is palindrome")
else:
    print("It is not a palindrome")
```
