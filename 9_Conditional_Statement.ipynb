# Conditional Statements

Conditional statements in Python are used for decision-making. They check whether a condition is True or False and execute the appropriate block of code. Python provides if, if-else, if-elif-else, and nested if statements for implementing conditions.  
  
Types of Conditional Statements:-  
1. if - Executes if condition if true
2. if-else - Executes if block if condition is True, otherwise else block will be executed
3. if-elif-else - Execute if block if condition is True, if not elif block will be executed otherwise else block will be executed.
4. Nested if - if block contains another if block

```python
# if
a=10
if a<15:
    print("It is less than 15")
```

```python
# if - else:
a=10
if a<9:
    print("Smaller than 9")
else:
    print("Larger than 9")
```

```python
# if-elif-else
a=int(input("Enter a number: "))   #taking input from user

print(a)   #printing the input just for checking(not mandatory)

if 0<=a<=9:
    print("It is single digit number")
elif 10<=a<=99:
    print("It is two digit number")
else:
    print("It is larger than 99")
```

We can use n numbers of elif blocks.

```python
# Nested if

a=int(input("Enter a number: "))   #taking input from user
print(a)

if 0<=a<=9:
    print("It is single digit number")
    if 5<=a<=7:
        print("It is between 5 and 7")
elif 10<=a<=99:
    print("It is two digit number")
else:
    print("It is larger than 99")
```

```python
## Accept 2 numbers and print the greatest betweem them

a=int(input("Enter 1st number: ")) 
b=int(input("Enter 2nd number: ")) 

if a>b:
    print(f"{a} is greater than {b}")
else:
    print(f"{b} is greater than {a}")
```

```python
## Accept the gender from user as Char and print the respective greeting message

gender = input("Enter gender as M/F: ")

if gender=='F':
    print("Goodmorning Ma'am")
elif gender == 'M':
    print("Goodmorning Sir")
else:
    print("Enter valid gender")
```

```python
## Accept an integer and check whether it is an even number or odd
n=int(input("Enter 1st number: "))

if n%2==0:
    print("It is even")
else:
    print("It is odd")
```

```python
## Accept name and age from the user. Check if the user is valid voter or not.

name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age>=18:
    print(f"Hello {name} you are a valid voter")
else:
    print(f"Hello {name} you are not a valid voter")
```

```python
## Accept a year and check if it is a leap year or not.
"""
Conditions for leap year:
1. Divisible by 400
or
2. Divisible by 4 but not by 100
"""

year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("It is a leap year")
else:
    print("It is not a leap year")
```
