# String

String is a data type that is used to store a sequence of characters, such as letters, number, or symbols, enclosed in single quotes(' '), double quotes(" "), or triple quotes(''' ''' or """ """). String is a immutable data type.

```python
a = 'A'
print(ord(a))  ## ord() is used for checking Unicode of string. Output: 65

b=65
print(chr(b))  ##Output: A

```

## Indexing


We can select and access a small portion of string using Indexing. The selected string is called sub-string.
Indexing starts from 0.

```python
name = "Mohd Zaid"
print(name[0])
```

We also have negative indexing.

```python
print(name[-1])
```

## String Slicing

Cutting a small portion or sub-string from string

Syntax: str[start:end:steps]

```python
txt='Hello I am learning Python'

print(txt[0:8:1])
```

In slicing, last count is not included. If we are giving range from 0 to 4, it will print characters from index 0 to 3.

```python
print(txt[20::1])
```
