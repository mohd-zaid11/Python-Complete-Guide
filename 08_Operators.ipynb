# Operators

Operators are symbols that perform operations on variables and values. Python has several types of operators for different tasks like arithmetic, comparison, logical operations, and more.  
Lets see every operator one by one.

## Arithmetic Operators

These operators perform mathematical operations like addition, subtraction, multiplication, division, etc.  
There are 7 types of arithmetic operators:-  
    1. addition         :       +  
    2. subtraction      :       -  
    3. multiplication   :       *   
    4. division         :       /  
    5. floor division   :       //  
    6. modulus          :       %  
    7. exponentiation   :       **  

```python
a=10
b=8

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
print(5/2*4-2+1)
```

## Assingment Operator

Used to assign values to operators.  
A basic assignment operator is =.

```python
a = 10    #we are assigning 10 to variable a
```

## Compund Assignment Operator

Compound assignment operator combines arithmetic operations with assignment.  
1. += Add and assign
2. -= Subtract and assign
3. *= Multiply and assign
4. /= Divide and assign
5. //= Flood divide and assign
6. %= Moduluz and assign
7. **= Exponentiation adn assign

```python
a = 10
a+=2
print(a)
a-=3
print(a)
a**=2
print(a)
a%=10
print(a)
```

# Comparision Operator

Also called Relational Operators, are used to compare two values.  
These operators will always provide Boolean result that is True or False.  
1. == Equal to
2. != Not equal to
3. '>' Greater than
4. < Less than
5. '>=' Greater than or equal to
6. <= Less than or equal to

These operators works with numbers but we can also use them with string as well.  
String will be comparing the Ascii values of string.  


```python
a=10
b=23
print(a==b)
print(a<=b)
print(a>b)
```

```python
a='A'
b='c'

print(a==b)
```

```python
a='2'
b='15'

print(a<b)   #here python interpretor will only check Aschii value of '1' from '15' and with compare it with Aschii value of '2'
```

## Logical Operator

These operators in Python are used to combine multiple conditions and return a Boolean result(True or False).  
3 type of Logical Operators:-  
1. and 
2. or
3. not

```python
print(15>10 and 20>26)
```

```python
print(15>10 or 20>26)
```

```python
print(bool(0))
print(not 0)
```

```python
print(not 10 == 10)
```
