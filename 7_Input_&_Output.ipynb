# Input & Output in Python

## Output

You probably know till now how to provide the output of the code you have written and that is with print() function.  
There is no other function to provide the result on the terminal we just have to use print() function.  
Now when providing the output we can use variables in print statement using a formatted string as shown in below example

```python
name = 'Xyz'
print(f"Hello {name}, let's master Python together")   ##this is formatted string(f" ")
```

## Input

It means asking user to provide a input that will be use in code.  
We can easily take input using input() function.  
Input usually take data in string by default. We use type conversion for converting input data into required data type.

```python
name = input("Enter your name: ")      #here we are taking an input and storing in name variable
print(f"Hello {name}")
```

```python
age = input('Enter your age: ')
print(age)
print(type(age))   #age is storing as string here while we gava int input
```

```python
age = int(input("Enter your age: "))    #now the input will be integer
print(age)
print(type(age))
```
