# Functions

- A function is a reusable block of code that performs a specific task. It helps avoid code repetition and makes programs more organized and modular.  
  
There are many built-n function like print(), input(), len(), etc.  
But we can also create our own functions and they are called user-defined functions.  

- To make your own function you have to use def keyword and then name the function.

```python
""" 
Function Syntax:-

def function_name(parameters):
    #Function body
    return value
"""
```

```python
def pehla_function():
    print("Hello hello helloo")

pehla_function()
```

```python
def sum(a,b):      #now it is compulsory to give values of a and b while calling the function otherwise it will through error
    print("The sum of your number is", a+b)

sum(12,13)
sum(10,4)
```

#### Types of Parameters:  
1. Positional Arguments
2. Keyword Arguments
3. Hoe Default Arguments

```python
## 1. Positional Arguments
def add(a,b):
    print(a+b)
add(3,5)

## 2. Keyword Arguments
def intro(name,age):
    print(f"{name}'s age is {age}")
intro(age=21, name="Zaid")

## 3. Hoe Default Arguments
def greet(name='Zaid'):
    print(f"Hello {name}, Goodmorning")
greet()    # Zaid will be default value for name here
greet('Alice')    #I can also give another value for name
```

#### return() Function  
It sends a value back to the function caller. While, print() displays output on the screen.  
The returned value form return() function can be stored in a variable or used in expressions.

```python
def sum(a,b):
    return a+b

print(sum(2,3))
```
