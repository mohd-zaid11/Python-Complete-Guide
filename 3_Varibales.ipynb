# Variables

Variables are the container that we use to contain some value.  
No - This is not actually correct definition for variables in Python for interview prespective.  
  
    
Correct definition is:  
A varibale in Python is a name used to store a value in memory. The value can be changed during the execution of the program.  
  
    
      
Basically instead of directly storing a value, it stores a reference(or memory location) to an object in memory.


```python
name = 'Zaid'            # Here 'name' is a varibale here and 'Zaid' is a value
age = 21

#we can also print it
print(name)
print(age)
```

Rules for creating variables:-
1. You can not start name of variable with number. Ex- 2name = 'Zaid'
2. You can not space anywhere in the name of variable.
3. You should not use special character in variables except '_'. Special characters have their own special uses in python.  
  
Naming Conventions:-  
You can write variables in pyhton in 3 ways.
1. Camel case   - mohdZaid
2. Pascal case  - MohdZaid
3. Snake case   - mohd_zaid 

```python
myName = 'Zaid'     #Camel Case
MyName = 'Zaid'     #Pascal Case
my_name ='Zaid'     #Snake Case

print(myName)
print(MyName)
print(my_name)
```
