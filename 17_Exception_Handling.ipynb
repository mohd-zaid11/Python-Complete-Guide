# Exception Handling

### Error
* Error occur due to mistakes in the code that prevent it from running. These can be syntax error or logical error.

### Exceptions
* Exceptions are unexpected events or errors that occurs during the execution of a program, which fisrupts the normal flow of program.

### Some important Keywords
1. try : Wrap the block of code that might cause an exception.
2. except : Handle the execution if it occurs.
3. else : Run code only if no exception occurs.
4. finally : Run code no matter what, whatever there's an exception or not.
5. raise : Manually throw an exception.
