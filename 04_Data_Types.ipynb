# Data Types

A Data Type is the type of data that tells Python what kind of value a variable stores and what operations can be performed.  
A Data Type defines the kind of value a variable holds.

```python
a=10
print(type(a))
```

Types of Data Types:-
1. Numbers  
    i. int -> Whole numbers  
    ii. float -> Decimal Numbers  
    iii. Complex -> Combination of real & imaginary parts  

2. String(str) -> Text

3. Boolean(bool) -> True/False

4. List(list) -> Collection of items

5. Tuple(tuple) -> Ordered, immutable collection

6. Set(set) -> Unordered collection of unique items

7. Dictionary(dict) -> Key-Value pairs  

To check data type of a variable we use type() function


```python
name = 'Xyz'
print(type(name))
```

```python
a=10/2
print(type(a))
```
