# Type Conversion

There are two types of Type Conversion:-  
1. Implicit
2. Explicit

In Implicit, Python automatically convert data type into required one.  
In Explicit, we have to manually convertion by assigning data types. Like:- int(), float(), str(), bool(), etc.

```python
a=12
print(type(a))
```

```python
a=str(a)
print(type(a))
```

```python
a=float(a)
print(type(a))
```

```python
b='10'
print(type(b))
```

```python
b=int(a)
print(type(b))
```

If we want to convert a value into int or float, the vlue must be numeric.

```python
a=10
a=bool(a)
print(type(a))
print(a)
```
