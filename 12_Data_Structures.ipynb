# Data Structures

- Data Structures are used to store, organize, and manipulate data eficiently. Python provides several built-in data structures.  
- In python, we have 4 types of in-build structures-
1. List
2. Tuple
3. Dictionary
4. Set  

There are some custom data structures as well like Stack, Queue, Linked List, Graph, etc. But these are the parts of Data Structure and Algorithm. We will only focus here on built-in data structures.
